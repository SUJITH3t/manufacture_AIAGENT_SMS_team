# Notebook 03: LLM Provider Abstraction & Latency Comparison

Evaluates the multi-provider routing architecture across Ollama, Groq, and Local Fallback.


```python
import sys
import os
import time
sys.path.insert(0, os.path.abspath('..'))

from backend.app.providers.router import ProviderRouter, get_llm_provider

router = ProviderRouter()

# Check availability
for p_name in ['fallback', 'ollama', 'groq']:
    prov = router.get_provider(p_name)
    avail = prov.is_available()
    print(f'Provider: {p_name:<10} | Name: {prov.get_provider_name():<25} | Available: {avail}')
```

## Test Inference Latency on Benchmark Prompt


```python
test_prompt = 'Analyze machine spindle temperature 78.5C with vibration 2.8 mm/s.'
for p_name in ['fallback']:
    prov = router.get_provider(p_name)
    t0 = time.time()
    response = prov.generate(test_prompt)
    dur = time.time() - t0
    print(f'Provider [{p_name}] Latency: {dur:.4f}s')
    print(f'Response excerpt:\n{response[:250]}...\n')
```
