Executes automated benchmark evaluation over the 24 curated test cases and visualizes accuracy, escalation precision, and latency distributions.


```python
import sys
import os
import pandas as pd
sys.path.insert(0, os.path.abspath('..'))

from evaluation.run_evaluation import run_benchmark

# Execute benchmark
run_benchmark('../evaluation/evaluation_dataset.json', provider='fallback')

# Load and display generated results dataframe
df_results = pd.read_csv('../evaluation/evaluation_results.csv')
df_results.head(10)
```

## Evaluation Summary Table by Category


```python
summary = df_results.groupby('category').agg({
    'risk_match': 'mean',
    'escalation_match': 'mean',
    'evidence_count': 'mean',
    'latency_sec': 'mean'
})
summary['risk_accuracy_%'] = summary['risk_match'] * 100
summary['escalation_accuracy_%'] = summary['escalation_match'] * 100
summary[['risk_accuracy_%', 'escalation_accuracy_%', 'evidence_count', 'latency_sec']]
```
